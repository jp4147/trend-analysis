```python
import pickle
from stella_emb_trend import Generate_Embeddings_Sequence

# Load the noise dictionary
file_path = r'R:\Kevin\Projects\trend-analysis\datasets\noise0.6_beta0.1.pickle'
with open(file_path, 'rb') as f:
    noise = pickle.load(f)

# Generate embeddings
generator = Generate_Embeddings_Sequence(noise, model_name="dunzhang/stella_en_1.5B_v5")
embeddings = generator.emb

# Save the embeddings
output_path = r'R:\Kevin\Projects\trend-analysis\embeddings\noise0.6_beta0.1_stella_emb.pickle'
with open(output_path, 'wb') as f:
    pickle.dump(embeddings, f)
```
